```python
import random

# 1. Define the Problem: Distance matrix for 5 cities
dist = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 17],
    [15, 35, 0, 30, 20],
    [20, 25, 30, 0, 15],
    [25, 17, 20, 15, 0]
]

n = len(dist)

# 2. Initialize Parameters
POP_SIZE = 10
MUTATION_RATE = 0.1
GENERATIONS = 100


# 3. Create Initial Population
def create_route():
    return random.sample(range(n), n)


population = [create_route() for _ in range(POP_SIZE)]


# 4. Evaluate Fitness
def total_distance(route):
    distance = 0

    for i in range(n):
        distance += dist[route[i]][route[(i + 1) % n]]

    return distance


def fitness(route):
    return 1 / total_distance(route)


# 5. Selection
def selection(population):
    return min(population, key=total_distance)


# 6. Crossover
def crossover(parent1, parent2):
    start, end = sorted(random.sample(range(n), 2))

    child = [-1] * n
    child[start:end] = parent1[start:end]

    remaining = [city for city in parent2 if city not in child]

    j = 0
    for i in range(n):
        if child[i] == -1:
            child[i] = remaining[j]
            j += 1

    return child


# 7. Mutation
def mutation(route):
    route = route.copy()

    if random.random() < MUTATION_RATE:
        i, j = random.sample(range(n), 2)
        route[i], route[j] = route[j], route[i]

    return route


# 8. Iteration
best_route = None

for generation in range(GENERATIONS):
    new_population = []

    best = selection(population)
    new_population.append(best.copy())

    while len(new_population) < POP_SIZE:
        parent1 = selection(population)
        parent2 = random.choice(population)

        child = crossover(parent1, parent2)
        child = mutation(child)

        new_population.append(child)

    population = new_population

    current_best = selection(population)

    if best_route is None or total_distance(current_best) < total_distance(best_route):
        best_route = current_best.copy()


# 9. Output the Best Solution
print("Best Route:", [city + 1 for city in best_route] + [best_route[0] + 1])
print("Minimum Distance:", total_distance(best_route))
```
