```python

import random

# Knapsack data
weights = [2, 3, 4, 5, 9]
profits = [3, 4, 5, 8, 10]
capacity = 10

# Cuckoo Search parameters
n_nests = 10
max_iterations = 100
pa = 0.25

# Generate random nest (0 or 1)
def create_nest():
    return [random.randint(0, 1) for _ in weights]

# Calculate fitness
def fitness(nest):
    total_weight = sum(nest[i] * weights[i] for i in range(len(weights)))
    total_profit = sum(nest[i] * profits[i] for i in range(len(profits)))

    if total_weight > capacity:
        return 0

    return total_profit

# Initialize nests
nests = [create_nest() for _ in range(n_nests)]

# Find best nest
best_nest = max(nests, key=fitness)

# Main Cuckoo Search loop
for iteration in range(max_iterations):

    for i in range(n_nests):

        # Generate a new solution
        new_nest = nests[i].copy()

        # Change random item selection
        position = random.randint(0, len(weights) - 1)
        new_nest[position] = 1 - new_nest[position]

        # Compare fitness
        if fitness(new_nest) > fitness(nests[i]):
            nests[i] = new_nest

    # Abandon some worst nests
    for i in range(n_nests):
        if random.random() < pa:
            nests[i] = create_nest()

    # Update best solution
    current_best = max(nests, key=fitness)

    if fitness(current_best) > fitness(best_nest):
        best_nest = current_best.copy()

# Display result
total_weight = sum(best_nest[i] * weights[i] for i in range(len(weights)))
total_profit = fitness(best_nest)

print("Best Solution:", best_nest)
print("Selected Items:", [i + 1 for i in range(len(weights)) if best_nest[i] == 1])
print("Total Weight:", total_weight)
print("Maximum Profit:", total_profit)
```
