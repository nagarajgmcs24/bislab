```python

import numpy as np

# Sphere function
def sphere(x):
    return np.sum(x ** 2)

# PSO parameters
n_particles = 30
dimensions = 2
max_iterations = 10

w = 0.7
c1 = 1.5
c2 = 1.5

# Initialize positions and velocities
positions = np.random.uniform(-10, 10,
                              (n_particles, dimensions))
velocities = np.random.uniform(-1, 1,
                               (n_particles, dimensions))

# Initialize personal best
pbest_positions = positions.copy()
pbest_values = np.array([sphere(x) for x in positions])

# Initialize global best
best_idx = np.argmin(pbest_values)
gbest_position = pbest_positions[best_idx].copy()
gbest_value = pbest_values[best_idx]

# PSO main loop
for iteration in range(max_iterations):

    # Evaluate particles and update personal best
    for i in range(n_particles):
        fitness = sphere(positions[i])

        if fitness < pbest_values[i]:
            pbest_values[i] = fitness
            pbest_positions[i] = positions[i].copy()

    # Update global best
    best_idx = np.argmin(pbest_values)

    if pbest_values[best_idx] < gbest_value:
        gbest_value = pbest_values[best_idx]
        gbest_position = pbest_positions[best_idx].copy()

    # Generate random values
    r1 = np.random.rand(n_particles, dimensions)
    r2 = np.random.rand(n_particles, dimensions)

    # Update velocity
    velocities = (
        w * velocities
        + c1 * r1 * (pbest_positions - positions)
        + c2 * r2 * (gbest_position - positions)
    )

    # Update position
    positions = positions + velocities

    print(f"Iteration {iteration + 1}: Best Fitness = {gbest_value:.5f}")

# Final result
print(f"Best Position: {gbest_position}")
print(f"Minimum Fitness: {gbest_value:.5f}")
```
